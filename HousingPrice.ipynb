Identifying columns and dropping it

```python
def toDrop(fileName,colsToRemove):
    for c in colsToRemove:
        fileName.drop(c,axis=1,inplace=True)

colsToRemove=['Id','Alley','MasVnrType','FireplaceQu','PoolQC','Fence','MiscFeature']
toDrop(data,colsToRemove)
toDrop(test,colsToRemove)
```

Filling the missing value with most occuring values

```python
def toUpdateMode(fileName,cols):
    for c in cols:
        fileName[c]=fileName[c].fillna(fileName[c].mode()[0])

dataColToUpdateMode=['BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1','BsmtFinType2','Electrical','GarageType','GarageFinish','GarageQual','GarageCond']
testColToUpdateMode=['BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1','BsmtFinType2','Electrical','GarageType','GarageFinish','GarageQual','GarageCond','MSZoning','Utilities','Functional','Exterior1st','Exterior2nd','KitchenQual','SaleType']
toUpdateMode(data,dataColToUpdateMode)
toUpdateMode(test,testColToUpdateMode)
```

```python
def toUpdateMean(fileName,cols):
    for c in cols:
        fileName[c]=fileName[c].fillna(fileName[c].mean())

dataColToUpdateMode=['LotFrontage','GarageYrBlt','MasVnrArea']
testColToUpdateMode=['LotFrontage','GarageYrBlt','MasVnrArea','BsmtFullBath','BsmtHalfBath','BsmtFinSF1','BsmtFinSF2','BsmtUnfSF','TotalBsmtSF','GarageCars','GarageArea']
toUpdateMean(data,dataColToUpdateMode)
toUpdateMean(test,testColToUpdateMode)
```

```python
from sklearn import preprocessing
le=preprocessing.LabelEncoder()
columnToEncode=['MSZoning',
 'Street',
 'LotShape',
 'LandContour',
 'Utilities',
 'LotConfig',
 'LandSlope',
 'Neighborhood',
 'Condition1',
 'Condition2',
 'BldgType',
 'HouseStyle',
 'RoofStyle',
 'RoofMatl',
 'Exterior1st',
 'Exterior2nd',
 'ExterQual',
 'ExterCond',
 'Foundation',
 'BsmtQual',
 'BsmtCond',
 'BsmtExposure',
 'BsmtFinType1',
 'BsmtFinType2',
 'Heating',
 'HeatingQC',
 'CentralAir',
 'Electrical',
 'KitchenQual',
 'Functional',
 'GarageType',
 'GarageFinish',
 'GarageQual',
 'GarageCond',
 'PavedDrive',
 'SaleType',
 'SaleCondition']
for col in columnToEncode:
    data[col]=le.fit_transform(data[col])
    test[col]=le.fit_transform(test[col])
```

```python
from sklearn.model_selection import train_test_split
y=data['SalePrice']
x=data.drop('SalePrice',axis=1)

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
lr.score(x_test,y_test)
```
